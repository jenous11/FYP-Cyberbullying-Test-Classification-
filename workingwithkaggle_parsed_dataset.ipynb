importing nltk libraries for text stopword removal

```python
import nltk

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

Modifyig the regex function by adding stpword removal from nltk

```python

def clean_text(data):
  if data is None:
        return ""

  result0 = re.sub(r'x[a-z0-9\s]+[\d]+'," ",data)
  # checking
  result = re.sub(r'https?[\S]+', " ", result0)
  partiallycleaned_result=result
  partiallycleaned_result
  #checking
  fully_cleaned_result=re.sub(r' @[a-zA-Z0-9]+'," ",partiallycleaned_result)
  fully_cleaned_result
  #remove special characters
  fully_cleaned_result = re.sub(r'\W+'," ",fully_cleaned_result)
  cleaned_data = fully_cleaned_result.strip().lower()
  # return cleaned_data
  # print(f"No stopword removal : {cleaned_data}")

  #getting set of english stopwords for efficient look up
  stop_words = set(stopwords.words('english'))
  word_tokens = [word for word in word_tokenize(cleaned_data)]
  filtered_sentence = [word for word in word_tokens if word not in stop_words]
  filtered_text = " ".join(filtered_sentence)

  #returnes filtered data with no stopwords
  # print(f"After stopword removal : {filtered_text}")
  return filtered_text
  #passing data to the function

# data="@ShreyaBafna3 You are SO stupid!! Check http://somelink.com \xc2\xa0"
# data="you are so stupid and the world is a bad place"
# clean_text(data)
```
